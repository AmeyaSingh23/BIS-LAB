```python
import random

POPULATION_SIZE = 10
GENERATIONS = 30
MUTATION_RATE = 0.08

TARGET_DIFFICULTY = 70.37

def create_chromosome():
    return [
        random.randint(50, 120),
        random.randint(5, 25),
        random.randint(2, 8),
        random.randint(3, 12)
    ]

def calculate_difficulty(chromosome):
    hp = chromosome[0]
    damage = chromosome[1]
    speed = chromosome[2]
    count = chromosome[3]

    hp_score = hp / 150
    damage_score = damage / 30
    speed_score = speed / 10
    count_score = count / 15

    difficulty = (
        25 * hp_score +
        25 * damage_score +
        20 * speed_score +
        30 * count_score
    )

    return difficulty

def fitness(chromosome):
    actual_difficulty = calculate_difficulty(chromosome)
    error = TARGET_DIFFICULTY - actual_difficulty

    return 1 / (1 + error ** 2)

def selection(population):
    candidates = random.sample(population, 2)
    return max(candidates, key=fitness)

def crossover(parent1, parent2):
    point = random.randint(1, 3)
    return parent1[:point] + parent2[point:]

def mutation(chromosome):
    chromosome = chromosome.copy()

    if random.random() < MUTATION_RATE:
        chromosome[0] = random.randint(50, 120)

    if random.random() < MUTATION_RATE:
        chromosome[1] = random.randint(5, 25)

    if random.random() < MUTATION_RATE:
        chromosome[2] = random.randint(2, 8)

    if random.random() < MUTATION_RATE:
        chromosome[3] = random.randint(3, 12)

    return chromosome

random.seed(100)

population = [
    create_chromosome()
    for _ in range(POPULATION_SIZE)
]

best_solution = None

for generation in range(GENERATIONS):

    new_population = []

    best = max(population, key=fitness)
    new_population.append(best)

    while len(new_population) < POPULATION_SIZE:

        parent1 = selection(population)
        parent2 = selection(population)

        child = crossover(parent1, parent2)
        child = mutation(child)

        new_population.append(child)

    population = new_population

    current_best = max(population, key=fitness)

    if best_solution is None or fitness(current_best) > fitness(best_solution):
        best_solution = current_best.copy()

    if generation % 5 == 0:
        print(
            f"Generation {generation}: "
            f"Difficulty = {calculate_difficulty(current_best):.2f}, "
            f"Fitness = {fitness(current_best):.6f}"
        )

print("\n----- FINAL RESULT -----")

print("Best Chromosome:")
print(best_solution)

print("\nGame Parameters:")

print("Enemy HP      :", best_solution[0])
print("Enemy Damage  :", best_solution[1])
print("Enemy Speed   :", best_solution[2])
print("Enemy Count   :", best_solution[3])

print("\nTarget Difficulty  :", TARGET_DIFFICULTY)
print("Actual Difficulty  :", round(calculate_difficulty(best_solution), 2))
print("Fitness            :", round(fitness(best_solution), 6))
```
